4. Funny String
File: problem4.py

```python
def funny_string(s):
    r = s[::-1]

    for i in range(1, len(s)):
        if abs(ord(s[i]) - ord(s[i - 1])) != abs(
            ord(r[i]) - ord(r[i - 1])
        ):
            return "Not Funny"

    return "Funny"


q = int(input())

for _ in range(q):
    s = input().strip()
    print(funny_string(s))
```

7. Alphabet Rangoli
File: problem7.py

```python
def print_rangoli(n):
    import string

    alphabet = string.ascii_lowercase
    width = 4 * n - 3
    lines = []

    for i in range(n):
        left = alphabet[n - 1:n - i - 1:-1]
        right = alphabet[n - i - 1:n]
        letters = left + right
        line = "-".join(letters)
        lines.append(line.center(width, "-"))

    print("\n".join(lines + lines[-2::-1]))


n = int(input())
print_rangoli(n)
```

13. Sherlock and the Valid String
File: problem13.py

```python
from collections import Counter

s = input().strip()
char_counts = Counter(s)
freq_counts = Counter(char_counts.values())

if len(freq_counts) == 1:
    print("YES")

elif len(freq_counts) == 2:
    (f1, c1), (f2, c2) = freq_counts.items()

    if (f1 == 1 and c1 == 1) or (f2 == 1 and c2 == 1):
        print("YES")
    elif abs(f1 - f2) == 1:
        if (f1 > f2 and c1 == 1) or (f2 > f1 and c2 == 1):
            print("YES")
        else:
            print("NO")
    else:
        print("NO")

else:
    print("NO")
```

14. Anagram
File: problem14.py

```python
from collections import Counter

q = int(input())

for _ in range(q):
    s = input().strip()
    n = len(s)

    if n % 2 != 0:
        print(-1)
    else:
        first = Counter(s[:n // 2])
        second = Counter(s[n // 2:])

        changes = sum((first - second).values())
        print(changes)
```

16. Caesar Cipher
File: problem16.py

```python
n = int(input())
s = input().strip()
k = int(input())

k %= 26
result = []

for char in s:
    if 'a' <= char <= 'z':
        result.append(chr((ord(char) - ord('a') + k) % 26 + ord('a')))
    elif 'A' <= char <= 'Z':
        result.append(chr((ord(char) - ord('A') + k) % 26 + ord('A')))
    else:
        result.append(char)

print("".join(result))
```

17. Separate the Numbers
File: problem17.py

```python
def separate_numbers(s):
    if s.startswith("0"):
        print("NO")
        return

    for length in range(1, len(s) // 2 + 1):
        first = s[:length]
        current = int(first)
        result = first

        while len(result) < len(s):
            current += 1
            result += str(current)

        if result == s:
            print("YES", first)
            return

    print("NO")


q = int(input())

for _ in range(q):
    separate_numbers(input().strip())
```

19. Weighted Uniform Strings
File: problem19.py

```python
s = input().strip()
n = int(input())

weights = set()
previous = ""
current_weight = 0

for char in s:
    if char == previous:
        current_weight += ord(char) - ord('a') + 1
    else:
        current_weight = ord(char) - ord('a') + 1

    weights.add(current_weight)
    previous = char

for _ in range(n):
    query = int(input())
    print("Yes" if query in weights else "No")
```

20.Super Reduced String

```python
s = input().strip()
stack = []

for char in s:
    if stack and stack[-1] == char:
        stack.pop()
    else:
        stack.append(char)

result = "".join(stack)

if result:
    print(result)
else:
    print("Empty String")
```
