## Male and Female Visual and Non-Visual Analysis

```python
df_male = df.loc[df["gender"]=="Male"]
```

```python
df_male["height"].agg(["count","min","max","mean","median","std","skew","kurt"])
```

```python
df_female = df.loc[df["gender"]=="Female"]
```

```python
df_female["height"].agg(["count","min","max","mean","median","std","skew","kurt"])
```

```python
mu_normal_10000 = df["height"].mean()

sigma_normal_10000 = df["height"].std()

print(mu_normal_10000, sigma_normal_10000)

one_std_right = mu_normal_10000 + (1 * sigma_normal_10000)

one_std_left = mu_normal_10000 - (1 * sigma_normal_10000)

two_std_right = mu_normal_10000 + (2 * sigma_normal_10000)

two_std_left = mu_normal_10000 - (2 * sigma_normal_10000)

three_std_right = mu_normal_10000 + (3 * sigma_normal_10000)

three_std_left = mu_normal_10000 - (3 * sigma_normal_10000)
```

```python
fig,axs = plt.subplots(2,1,layout="constrained")
axs[0].axvline(mu_normal_10000, color='black', label='Mean')
axs[0].axvline(one_std_right, color='red', label='Mean + 1SD')
axs[0].axvline(one_std_left, color='red', label='Mean - 1SD')
axs[0].axvline(two_std_right, color='brown', label='Mean + 2SD')
axs[0].axvline(two_std_left, color='brown', label='Mean - 2SD')
axs[0].axvline(three_std_right, color='blue', label='Mean + 3SD')
axs[0].axvline(three_std_left, color='blue', label='Mean - 3SD')
axs[0].legend(loc ="best",bbox_to_anchor = (1,1))
sns.set_style("darkgrid")
sns.histplot(df_male["height"],kde=True,ax=axs[0])


axs[1].axvline(mu_normal_10000, color='black', label='Mean')
axs[1].axvline(one_std_right, color='red', label='Mean + 1SD')
axs[1].axvline(one_std_left, color='red', label='Mean - 1SD')
axs[1].axvline(two_std_right, color='brown', label='Mean + 2SD')
axs[1].axvline(two_std_left, color='brown', label='Mean - 2SD')
axs[1].axvline(three_std_right, color='blue', label='Mean + 3SD')
axs[1].axvline(three_std_left, color='blue', label='Mean - 3SD')
axs[1].legend(loc = "best",bbox_to_anchor = (1,1))
sns.set_style("darkgrid")
sns.histplot(df_female["height"],kde=True,ax=axs[1])
```
