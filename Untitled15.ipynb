Section A: CRUD Operations on Series

```python
import pandas as pd

# Create
marks = pd.Series(
    [90, 85, 95, 88],
    index=["Math", "Science", "English", "Physics"]
)

print(marks)
```

Read
2. Display Science marks

```python
print(marks["Science"])
```

3. Display all subjects and marks

```python
print(marks)
```

Update
4. Update Science marks

```python
marks["Science"] = 90
print(marks)
```

5. Increase all marks by 5

```python
marks = marks + 5
print(marks)
```

Delete
6. Remove Physics

```python
marks = marks.drop("Physics")
print(marks)
```

Analysis
7. Highest marks

```python
print(marks.max())
```

8. Lowest marks

```python
print(marks.min())
```

9. Average marks

```python
print(marks.mean())
```

10. Total number of subjecTS

```python
print(marks.count())
```

Section B: DataFrame CRUD

```python
data = {
    "Emp_ID": [101, 102, 103],
    "Name": ["Rahul", "Priya", "Arjun"],
    "Salary": [50000, 60000, 70000]
}

df = pd.DataFrame(data)

print(df)
```

Section C: Row-Level CRUD
Create Rows
1. Add one employee

```python
df.loc[len(df)] = [104, "Neha", 55000]

print(df)
```

```python
new_employees = pd.DataFrame({
    "Emp_ID": [105, 106],
    "Name": ["Kiran", "Asha"],
    "Salary": [65000, 48000]
})

df = pd.concat([df, new_employees], ignore_index=True)

print(df)
```

```python
print(df)
```

```python
print("Before:", df.shape)
```

```python
print("After:", df.shape)
```

Read Rows
1. First row

```python
print(df.iloc[0])
```

```python
print(df.iloc[-1])
```

```python
print(df.iloc[:3])
```

```python
print(df.loc[0])
```

```python
print(df.iloc[0])
```

```python
print(df.iloc[::-1])
```

Update Rows
1. Update salary of Employee 102

```python
df.loc[df["Emp_ID"] == 102, "Salary"] = 65000

print(df)
```

```python
df.loc[df["Emp_ID"] == 102, ["Name", "Salary"]] = ["Priya Reddy", 70000]

print(df)
```

```python
df["Salary"] = df["Salary"] * 1.10

print(df)
```

```python
df.loc[df["Salary"] > 60000, "Salary"] += 5000

print(df)
```

Delete Rows
1. Delete Employee ID 103

```python
df = df[df["Emp_ID"] != 103]

print(df)
```

```python
df = df[~df["Emp_ID"].isin([104, 105])]

print(df)
```

```python
df = df[df["Salary"] >= 50000]

print(df)
```

```python
print("Current shape:", df.shape)
```
