```python
#Q7
text = "Python 123!"
vowels = "aeiou"
count = 0

for char in text.lower():
    if char.isalpha() and char not in vowels:
        count += 1

print(count)
```

```python
#Q8
text = "banana".replace(" ", "")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

most_frequent = text[0]

for char in text:
    if frequency[char] > frequency[most_frequent]:
        most_frequent = char

print(most_frequent)
```

```python
#Q10
text = "programming".replace(" ", "")
frequency = {}

for char in text:
    frequency[char] = frequency.get(char, 0) + 1

repeated = []

for char in text:
    if frequency[char] > 1 and char not in repeated:
        repeated.append(char)

print(repeated)
```
