11. String Formatting
File: problem11.py

```python
n = int(input())

width = len(bin(n)) - 2

for i in range(1, n + 1):
    print(
        str(i).rjust(width),
        oct(i)[2:].rjust(width),
        hex(i)[2:].upper().rjust(width),
        bin(i)[2:].rjust(width)
    )
```

12. Set .discard(), .remove() & .pop()
File: problem12.py

```python
n = int(input())
s = set(map(int, input().split()))
number_of_commands = int(input())

for _ in range(number_of_commands):
    command = input().split()

    if command[0] == "pop":
        s.pop()
    elif command[0] == "remove":
        s.remove(int(command[1]))
    elif command[0] == "discard":
        s.discard(int(command[1]))

print(sum(s))
```

17. Check Strict Superset
File: problem17.py

```python
a = set(map(int, input().split()))
n = int(input())

result = True

for _ in range(n):
    other = set(map(int, input().split()))

    if not a.issuperset(other) or a == other:
        result = False
        break

print(result)
```

18. Set Mutations
File: problem18.py

```python
n = int(input())
a = set(map(int, input().split()))

number_of_operations = int(input())

for _ in range(number_of_operations):
    command, size = input().split()
    other = set(map(int, input().split()))

    if command == "update":
        a.update(other)
    elif command == "intersection_update":
        a.intersection_update(other)
    elif command == "difference_update":
        a.difference_update(other)
    elif command == "symmetric_difference_update":
        a.symmetric_difference_update(other)

print(sum(a))
```

19. Company Logo
File: problem19.py

```python
from collections import Counter

s = input().strip()
counts = Counter(s)

result = sorted(counts.items(), key=lambda item: (-item[1], item[0]))

for char, count in result[:3]:
    print(char, count)
```

20. Validating the Phone Number
File: problem20.py

```python
import re

n = int(input())

for _ in range(n):
    phone = input().strip()

    if re.fullmatch(r"[789]\d{9}", phone):
        print("YES")
    else:
        print("NO")
```
