Q1. What is Pandas? Explain its Features and Applications in Data Science.
1. What is Pandas?

Pandas is an open-source Python library used for data manipulation, data analysis, and data cleaning. It provides powerful data structures such as Series and DataFrame to organize, process, and analyze structured data efficiently.

Pandas is widely used in data science because it makes working with large datasets easier.

````python
import pandas as pd

data = {
    "Name": ["Ravi", "Priya", "Arjun"],
    "Age": [22, 25, 23],
    "Salary": [25000, 30000, 28000]
}

df = pd.DataFrame(data)

print(df)
````

Features of Pandas
1. Data Structures: Provides two main data structures: Series (one-dimensional) and DataFrame (two-dimensional).
2. Data Cleaning: Helps handle missing values, duplicate records, and incorrect data.
3. Data Manipulation: Allows filtering, sorting, selecting, and modifying data.
4. Data Analysis: Provides statistical functions such as mean, median, sum, minimum, and maximum.
5. Data Handling: Supports file formats such as CSV, Excel, JSON, and SQL data sources.
6. Data Grouping: Groups data using groupby() to perform calculations.
7. Data Merging: Combines datasets using merge(), join(), and concat().
8. Time Series Analysis: Supports working with dates, times, and time-based data.
9. Missing Value Handling: Provides functions such as isnull(), fillna(), and dropna().
10. Integration: Works well with NumPy, Matplotlib, Seaborn, and other Python libraries.

Applications of Pandas in Data Science
1. Data Cleaning: Removes duplicates and handles missing values.
2. Exploratory Data Analysis (EDA): Helps understand data using summaries and statistics.
3. Data Visualization: Prepares data for charts and graphs.
4. Business Analysis: Analyzes sales, revenue, profit, and customer behavior.
5. Financial Analysis: Processes stock prices, transactions, and financial records.
6. Machine Learning: Prepares datasets before training machine learning models.
7. Healthcare Analysis: Analyzes patient records and medical datasets.
8. Time Series Analysis: Examines trends in stock prices, sales, and weather data.
9. Report Generation: Summarizes data and generates useful reports.
10. Data Integration: Combines information from multiple files and databases.

4. Advantages of Pandas
- Easy to learn and use.
- Saves time when processing data.
- Handles missing and duplicate values.
- Supports multiple data formats.
- Provides powerful data analysis functions.
- Integrates with other Python libraries.

Q3: Create a Series from a Python List
Definition
In Pandas, a Series is a one-dimensional labeled array that can store data such as integers, strings, and floating-point numbers.

````python
import pandas as pd

# Create a Python list
data = [10, 20, 30, 40, 50]

# Convert the list into a Pandas Series
s = pd.Series(data)

# Display the Series
print(s)
````

Q4: Create a Series from a NumPy Array with Custom Indexes
Definition
In Pandas, we can create a Series from a NumPy array using pd.Series(). We can specify custom indexes using the index parameter instead of the default numerical indexes.

````python
import numpy as np
import pandas as pd

# Create a NumPy array
arr = np.array([10, 20, 30, 40, 50])

# Create a Series with custom indexes
s = pd.Series(arr, index=['a', 'b', 'c', 'd', 'e'])

# Display the Series
print(s)
````

Q20. Find the shape, dimensions, total elements, and datatype of a DataFrame using shape, ndim, size, and type().
Answer:

In Pandas, we use shape, ndim, and size to inspect a DataFrame. The Python type() function identifies the object's data type.

````python
#```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Rani', 'Raj', 'Priya'],
    'Age': [21, 22, 20, 21],
    'Marks': [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Shape:", df.shape)
print("Dimensions:", df.ndim)
print("Total elements:", df.size)
print("Data type:", type(df))
````

Q21. Display complete information using .info() and .describe().
Answer:

In Pandas, we use the info() method to display the structure and information of a DataFrame. The describe() method provides statistical information about numerical columns.

````python
#```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Rani', 'Raj', 'Priya', 'Kiran'],
    'Age': [21, 22, 20, 21, 23],
    'Marks': [85, 90, 78, 95, 88]
}

df = pd.DataFrame(data)

print("Complete Information:")
df.info()

print("\nStatistical Information:")
print(df.describe())
#```
````

Q22. Display dtypes and empty in Pandas.
Answer:

In Pandas, dtypes displays the data type of each column, while empty checks whether a DataFrame has any rows or columns.

````python
#```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Rani', 'Raj'],
    'Age': [21, 22, 20],
    'Marks': [85, 90, 78]
}

df = pd.DataFrame(data)

print("Data Types:")
print(df.dtypes)

print("\nIs DataFrame Empty?")
print(df.empty)
#```
````

Q23. Display three random rows using .sample().
Answer:

In Pandas, we use the sample() method to select random rows from a DataFrame. To display three random rows, we use df.sample(n=3).

````python
#```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Rani', 'Raj', 'Priya', 'Kiran'],
    'Age': [21, 22, 20, 21, 23],
    'Marks': [85, 90, 78, 95, 88]
}

df = pd.DataFrame(data)

print(df.sample(n=3))
#```
````

Q24. Transpose the DataFrame using .T.
Answer:

In Pandas, we use the .T attribute to transpose a DataFrame. Transposing means converting rows into columns and columns into rows.

````python
#```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Rani', 'Raj'],
    'Age': [21, 22, 20],
    'Marks': [85, 90, 78]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

print("\nTransposed DataFrame:")
print(df.T)
````
