# Marginal Probabilities(individual column)

### what is the probability of getting rain tommorow

```python
df['Outlook'].value_counts(normalize=True)
```

### What is the probability of playing cricket tommorow?

```python
df['Play'].value_counts(normalize=True).loc['Yes']
```

### What is the Probability that temperature is not cool tommorow?

```python
sum(df['Temperature']=='Cool')/len(df) #cool
```

```python
1-sum(df['Temperature']=='Cool')/len(df) #notcool
```

### Joint Probabilites ( and | or) - Multiple Columns

### what is the probability of playing cricket and temparature is hot?

```python
pd.crosstab(index=df['Play'],columns=df['Temperature'])
```

```python
pd.crosstab(index=df['Play'],columns=df['Temperature'],normalize=True)
```

```python
pd.crosstab(index=df['Play'],columns=df['Temperature'],normalize=True).loc['Yes']['Hot']
```

### what is the probability of humidity is normal and temerature is cool?

```python
pd.crosstab(index=df['Humidity'],columns=df['Temperature'])
```

```python
pd.crosstab(index=df['Humidity'],columns=df['Temperature'],normalize=True)
```

```python
pd.crosstab(index=df['Humidity'],columns=df['Temperature'],normalize=True).loc['Normal']['Cool']
```

## Conditional Probability -- one is happend , what is the chance of other event

## what is probability of playing cricket if it is raining

```python
pd.crosstab(index=df['Play'],columns=df['Outlook'],normalize='columns')
```

```python
pd.crosstab(index=df['Play'],columns=df['Outlook'],normalize='columns').loc['Yes']['Rain']
```

## what is the probability of not playing cricket when humidity is high ?

```python
pd.crosstab(index=df['Humidity'],columns=df['Temperature'],normalize='index')
```

```python
pd.crosstab(index=df['Humidity'],columns=df['Temperature'],normalize='index')
```
