PART:B Boolean Filtering

```python
(5)
A = np.array([5, 12, 17, 3, 8, 20])

print("Values > 10:", A[A > 10])
print("Values <= 8:", A[A <= 8])
```

```python
(6)
A = np.array([10, 15, 20, 25, 30, 35])

result = A[(A > 15) & (A % 5 == 0)]

print(result)
```

```python
(7)
A = np.array([3, -2, 0, 5, -7, 8])

non_negative = A[A >= 0]
negative_count = np.sum(A < 0)

print("Non-negative values:", non_negative)
print("Negative count:", negative_count)
```

```python
(8) #boolean replacement
A = np.array([1, 2, 3, 4, 5, 6, 7])

A[A % 2 != 0] = -1

print(A)
```

PART D: Boolean Indexing vs np.where()

```python
(13) #filtering vs transforming
A = np.array([4, 9, 2, 7, 12, 3])

filtered = A[A > 5]
transformed = np.where(A > 5, 1, 0)

print("Boolean indexing:", filtered)
print("np.where:", transformed)
```

```python
(14) #replace negavites-two methods(1.Boolean indexing)
A = np.array([-5, 10, -3, 8, -1])

B = A.copy()
B[B < 0] = 0

print(B)
```

```python
(14) #(2.np.where)
A = np.array([-5, 10, -3, 8, -1])

B = np.where(A < 0, 0, A)

print(B)
```

```python
(15) #Labeling Data with np.where
scores = np.array([45, 60, 30, 85, 90])

filtered_scores = scores[scores >= 50]

labels = np.where(scores >= 50, "Pass", "Fail")

print("Filtered scores:", filtered_scores)
print("Labels:", labels)
```

PART E:Broadcasting

```python
(16) #Row-wise Broadcasting
M = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

A = np.array([10, 20, 30])

print("Addition:")
print(M + A)

print("Subtraction:")
print(M - A)
```

```python
(17) #Row Normalizatuon
M = np.array([
    [2, 4],
    [6, 8]
])

A = np.array([2, 2])

print(M / A)
```

```python
(18) #Scalar Broadcasting
A = np.array([10, 20, 30])

print(A + 5)
```

```python
(19) # Boolean vs np.where() on 2D
M = np.array([
    [5, 10, 15],
    [20, 25, 30]
])
```

```python
 #Boolean indexing
A = M.copy()
A[A > 20] = 20

print(A)
```

```python
# np.where()
B = np.where(M > 20, 20, M)

print(B)
```

```python
(20) #Broadcasting Concept Check
M = np.array([
    [1, 2],
    [3, 4],
    [5, 6]
])

A = np.array([10, 20])

result = M + A

print(result)
```
