```python
import pandas as pd

df = pd.DataFrame({
    "Emp_ID": [101,102,103,104,105,106,107,108],
    "Name": ["Rahul","Priya","Arjun","Sneha",
             "Kiran","Pooja","Vikram","Anjali"],
    "Department": ["DS","HR","IT","Finance",
                   "IT","HR","DS","Finance"],
    "City": ["Hyderabad","Bangalore","Chennai","Mumbai",
             "Hyderabad","Bangalore","Chennai","Mumbai"],
    "Salary": [50000,45000,60000,55000,
               70000,48000,75000,65000],
    "Experience": [2,4,5,3,8,4,9,6],
    "Rating": [4,3,5,4,5,3,5,4]
})

print(df)
```

Section A: Filtering
1. Basic Filtering

```python

# 1. Salary greater than 50000
print(df[df["Salary"] > 50000])

# 2. Experience greater than 5 years
print(df[df["Experience"] > 5])

# 3. Rating equal to 5
print(df[df["Rating"] == 5])

# 4. Employees from Hyderabad
print(df[df["City"] == "Hyderabad"])

# 5. Employees from IT department
print(df[df["Department"] == "IT"])
```

2. Multiple Condition Filtering

```python

# 1. IT employees with salary greater than 60000
print(df[(df["Department"] == "IT") & (df["Salary"] > 60000)])

# 2. DS employees with rating 5
print(df[(df["Department"] == "DS") & (df["Rating"] == 5)])

# 3. Salary greater than 50000 and rating greater than 4
print(df[(df["Salary"] > 50000) & (df["Rating"] > 4)])

# 4. Employees from Hyderabad or Chennai
print(df[(df["City"] == "Hyderabad") | (df["City"] == "Chennai")])
```

Section B: GroupBy
3. GroupBy Basics

```python

# 1. Group employees by Department
grouped = df.groupby("Department")
print(grouped)

# 2. Display all department groups
for department, group in grouped:
    print(department)
    print(group)

# 3. Total number of groups
print(grouped.ngroups)

# 4. Size of each group
print(grouped.size())
```

4. GroupBy Analysis

```python

# 1. Employee count department-wise
print(df.groupby("Department")["Emp_ID"].count())

# 2. Employee count city-wise
print(df.groupby("City")["Emp_ID"].count())

# 3. Total salary department-wise
print(df.groupby("Department")["Salary"].sum())

# 4. Average salary department-wise
print(df.groupby("Department")["Salary"].mean())

# 5. Minimum salary department-wise
print(df.groupby("Department")["Salary"].min())

# 6. Maximum salary department-wise
print(df.groupby("Department")["Salary"].max())

# 7. Average rating department-wise
print(df.groupby("Department")["Rating"].mean())
```

Section C: Aggregation Functions
5. Basic Aggregations

```python

# 1. Total salary
print(df["Salary"].sum())

# 2. Average salary
print(df["Salary"].mean())

# 3. Minimum salary
print(df["Salary"].min())

# 4. Maximum salary
print(df["Salary"].max())

# 5. Average experience
print(df["Experience"].mean())

# 6. Maximum rating
print(df["Rating"].max())
```

6. Department-wise Aggregations

```python

# 1. Total salary
print(df.groupby("Department")["Salary"].sum())

# 2. Average salary
print(df.groupby("Department")["Salary"].mean())

# 3. Minimum salary
print(df.groupby("Department")["Salary"].min())

# 4. Maximum salary
print(df.groupby("Department")["Salary"].max())

# 5. Average experience
print(df.groupby("Department")["Experience"].mean())
```

Section D: .agg() Method
7. Single Column Aggregation

```python

# 1. Sum
print(df["Salary"].agg("sum"))

# 2. Mean
print(df["Salary"].agg("mean"))

# 3. Minimum
print(df["Salary"].agg("min"))

# 4. Maximum
print(df["Salary"].agg("max"))

# 5. Count
print(df["Salary"].agg("count"))
```

8. Multiple Aggregations

```python

# 1. Sum and mean of Salary
print(df["Salary"].agg(["sum", "mean"]))

# 2. Min and max of Salary
print(df["Salary"].agg(["min", "max"]))

# 3. Count, mean and max of Experience
print(df["Experience"].agg(["count", "mean", "max"]))

# 4. Multiple salary aggregations department-wise
print(
    df.groupby("Department")["Salary"].agg(
        ["count", "sum", "mean", "min", "max"]
    )
)
```

Section E: Apply Function
9. Apply Function Practice

```python

# 1. Increase salary by 10%
df["Increased_Salary"] = df["Salary"].apply(lambda x: x * 1.10)

# 2. Calculate annual salary
df["Annual_Salary"] = df["Salary"].apply(lambda x: x * 12)

# 3. Calculate bonus (10%)
df["Bonus"] = df["Salary"].apply(lambda x: x * 0.10)

# 4. Round increased salary
df["Rounded_Salary"] = df["Salary"].apply(
    lambda x: round(x * 1.10)
)

# 5. Add 1 year to Experience
df["New_Experience"] = df["Experience"].apply(lambda x: x + 1)

# 6. Square of Experience
df["Experience_Square"] = df["Experience"].apply(lambda x: x ** 2)

print(df)
```

Section F: Lambda Functions
10. Lambda Practice

```python

# 1. Increase salary by 10%
df["New_Salary"] = df["Salary"].apply(lambda x: x * 1.10)

# 2. Annual Salary
df["Annual_Salary"] = df["Salary"].apply(lambda x: x * 12)

# 3. Bonus = 10% of Salary
df["Bonus"] = df["Salary"].apply(lambda x: x * 0.10)

# 4. Tax = 5% of Salary
df["Tax"] = df["Salary"].apply(lambda x: x * 0.05)

# 5. Add 1 year to Experience
df["Updated_Experience"] = df["Experience"].apply(lambda x: x + 1)

print(df)
```

Section G: Map Function
11. Mapping Practice

```python

# 1. Department Code
department_codes = {
    "DS": "D01",
    "HR": "H01",
    "IT": "I01",
    "Finance": "F01"
}

df["Department_Code"] = df["Department"].map(department_codes)

# 2. City Code
city_codes = {
    "Hyderabad": "HYD",
    "Bangalore": "BLR",
    "Chennai": "CHN",
    "Mumbai": "MUM"
}

df["City_Code"] = df["City"].map(city_codes)

print(df[["Name", "Department_Code", "City_Code"]])
```

Section H: Pivot
12. Pivot Operations
Because each department or city has multiple employees, a regular pivot() requires a unique employee identifier. We use Emp_ID as columns.

```python

# 1. Department vs Salary
print(df.pivot(
    index="Department",
    columns="Emp_ID",
    values="Salary"
))

# 2. Department vs Experience
print(df.pivot(
    index="Department",
    columns="Emp_ID",
    values="Experience"
))

# 3. City vs Salary
print(df.pivot(
    index="City",
    columns="Emp_ID",
    values="Salary"
))
```

Section I: Pivot Table
13. Pivot Table Operations

```python

# 1. Average salary by Department
print(pd.pivot_table(
    df,
    index="Department",
    values="Salary",
    aggfunc="mean"
))

# 2. Total salary by Department
print(pd.pivot_table(
    df,
    index="Department",
    values="Salary",
    aggfunc="sum"
))

# 3. Average salary by Department and City
print(pd.pivot_table(
    df,
    index="Department",
    columns="City",
    values="Salary",
    aggfunc="mean"
))

# 4. Total salary by Department and City
print(pd.pivot_table(
    df,
    index="Department",
    columns="City",
    values="Salary",
    aggfunc="sum"
))
```
