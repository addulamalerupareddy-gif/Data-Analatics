Phase 1: Dataset Creation
Task 1: Create NumPy arrays

```python

import numpy as np
import pandas as pd

ages = [25, 30, 35, 28, 40]
employee_ids = [101, 102, 103, 104, 105]
dept_ids = [1, 1, 2, 3, 2]
salaries = [50000, 60000, 70000, 55000, 65000]

age = np.array(ages)
employee_id = np.array(employee_ids)
dept_id = np.array(dept_ids)
salary = np.array(salaries)

# Create an array using arange()
arr_range = np.arange(100, 105)

# Five random integers between 20 and 50, inclusive
random_arr = np.random.randint(20, 51, size=5)

print("Age:", age)
print("Employee ID:", employee_id)
print("Department ID:", dept_id)
print("Salary:", salary)
print("Arange:", arr_range)
print("Random integers:", random_arr)
```

Task 2: Create Pandas Series

```python

age_series = pd.Series(age)
employee_id_series = pd.Series(employee_id)
dept_id_series = pd.Series(dept_id)
salary_series = pd.Series(salary)

name_series = pd.Series(
    ["Alice", "Bob", "Charlie", "David", "Eve"]
)

print(age_series)
print(name_series)
```

Task 3: Convert Series into a DataFrame

```python

employee_df = pd.DataFrame({
    "employee_id": employee_id_series,
    "name": name_series,
    "age": age_series,
    "dept_id": dept_id_series,
    "salary": salary_series
})

print(employee_df)
```

Phase 2: Department Dataset
Task 4: Create department Series

```python

dept_id_series = pd.Series([1, 2, 3])
department_name_series = pd.Series(["HR", "IT", "Finance"])
location_series = pd.Series(["Hyderabad", "Bangalore", "Mumbai"])

print(dept_id_series)
print(department_name_series)
print(location_series)
```

Task 5: Create department DataFrame

```python

department_df = pd.DataFrame({
    "dept_id": dept_id_series,
    "department_name": department_name_series,
    "location": location_series
})

print(department_df)
```

Phase 3: Combining Data
Task 6: Merge — inner join and left join

```python

# Inner join
inner_df = pd.merge(
    employee_df, department_df,
    on="dept_id", how="inner"
)
print("Inner Join:")
print(inner_df)

# Left join
df = pd.merge(
    employee_df, department_df,
    on="dept_id", how="left"
)
print("Left Join:")
print(df)
```

Task 7: Concatenation and bonus

```python

new_employees = pd.DataFrame({
    "employee_id": [106, 107],
    "name": ["Frank", "Grace"],
    "age": [32, 27],
    "dept_id": [1, 3],
    "salary": [58000, 52000]
})

# Add new employees row-wise
employee_df = pd.concat(
    [employee_df, new_employees],
    ignore_index=True
)

# Add bonus = 10% of salary
employee_df["bonus"] = employee_df["salary"] * 0.10

print(employee_df)
```

```python

df = pd.merge(
    employee_df, department_df,
    on="dept_id", how="left"
)

print(df)
```

Phase 4: Data Selection and Indexing
Task 8: Selecting Data

```python

# Select a single column
print(df["name"])

# Select multiple columns
print(df[["name", "age", "salary"]])

# First 3 rows
print(df.head(3))

# Rows at index 1 to 3 (inclusive)
print(df.loc[1:3])

# Show multiple columns
print(df[["name", "age", "salary"]])
```

Task 9: loc vs iloc

```python

# loc: label-based selection, includes index label 3
print(df.loc[1:3, ["name", "salary"]])

# iloc: position-based selection, end position excluded
print(df.iloc[1:4, [1, 4]])
```

Task 10: Advanced Indexing

```python

# Salary greater than 60000
print(df[df["salary"] > 60000])

# Fancy indexing: select rows by index labels
print(df.loc[[0, 2, 4]])

# Slice rows and columns
print(df.iloc[1:4, 1:5])
```

Task 11: Conditional Selection

```python

# Employees in HR or IT
print(df[df["department_name"].isin(["HR", "IT"])])

# Employees not in Finance
print(df[df["department_name"] != "Finance"])

# Replace salary with NaN where age < 30
df_conditional = df.copy()
df_conditional.loc[df_conditional["age"] < 30, "salary"] = np.nan

print(df_conditional)
```

Phase 5: Data Cleaning
Task 12: Missing Values

```python

# Introduce missing values in a copy
df_missing = df.copy()
df_missing.loc[0, "salary"] = np.nan
df_missing.loc[2, "age"] = np.nan

print("Before cleaning:")
print(df_missing)

# Count missing values
print("\nMissing values:")
print(df_missing.isnull().sum())

# Fill missing salary with mean
df_missing["salary"] = df_missing["salary"].fillna(
    df_missing["salary"].mean()
)

print("\nAfter filling salary:")
print(df_missing)

# Drop rows containing any null values
df_dropped = df_missing.dropna()
print("\nAfter dropping remaining nulls:")
print(df_dropped)
```

Task 13: Duplicates

```python

df_duplicates = df.copy()

# Duplicate the first row
df_duplicates = pd.concat(
    [df_duplicates, df_duplicates.iloc[[0]]],
    ignore_index=True
)

# Count duplicate rows
print("Duplicate count:", df_duplicates.duplicated().sum())

# Remove duplicates
df_duplicates = df_duplicates.drop_duplicates()

print(df_duplicates)
```

Task 14: Outlier Detection using IQR

```python

Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["salary"] < lower_bound) |
    (df["salary"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Salary outliers:")
print(outliers)
```

Phase 6: Data Transformation
Task 15: Create New Columns

```python

df_transform = df.copy()

# Bonus = 10% of salary
df_transform["bonus"] = df_transform["salary"] * 0.10

# Function to categorize age
def get_age_group(age):
    if age < 30:
        return "Young"
    elif age < 40:
        return "Mid"
    else:
        return "Senior"

df_transform["age_group"] = df_transform["age"].apply(get_age_group)

print(df_transform)
```

Task 16: Rename Columns

```python

# Rename employee_id
df_renamed = df.rename(
    columns={"employee_id": "emp_id"}
)
print(df_renamed.columns)

# Rename multiple columns
df_renamed = df.rename(columns={
    "dept_id": "department_id",
    "salary": "monthly_salary"
})

print(df_renamed.columns)
```

Task 17: Drop Operations

```python

# Drop dept_id
df_drop1 = df.drop(columns=["dept_id"])
print("After dropping dept_id:", df_drop1.shape)

# Drop multiple columns
df_drop2 = df.drop(columns=["location", "age"])
print("After dropping multiple columns:", df_drop2.shape)

# Drop row at index 2
df_drop3 = df.drop(index=2)
print("After dropping row:", df_drop3.shape)
```

Phase 7: Sorting and Indexing
Task 18: Sorting

```python

# Sort by salary descending
print(df.sort_values("salary", ascending=False))

# Department A-Z, salary descending within department
print(df.sort_values(
    ["department_name", "salary"],
    ascending=[True, False]
))

# Sort by index
print(df.sort_index())
```

Task 19: Index Management

```python

# Reset index
df_reset = df.reset_index(drop=True)
print(df_reset)

# Set employee_id as index
df_indexed = df.set_index("employee_id")
print(df_indexed)
```

Phase 8: Grouping and Aggregation
Task 20: GroupBy

```python

avg_salary = df.groupby("department_name")["salary"].mean()

print(avg_salary)
```

Task 21: Aggregations

```python

# Multiple aggregations on salary
print(df["salary"].agg(["count", "sum", "mean", "min", "max"]))

# Aggregations on salary and age
print(df.agg({
    "salary": ["count", "sum", "mean", "min", "max"],
    "age": ["count", "mean", "min", "max"]
}))
```

Task 22: idxmax() and idxmin()

```python

# Employee with highest salary
print(df.loc[df["salary"].idxmax()])

# Employee with lowest age
print(df.loc[df["age"].idxmin()])
```

Phase 9: Functional Operations
Task 23: Apply, Map, and Filter

```python

df_functions = df.copy()

# Apply a function to age: add 1 year
df_functions["age"] = df_functions["age"].apply(lambda x: x + 1)

# Map department names
department_map = {
    "HR": "Human Resources",
    "IT": "Information Technology",
    "Finance": "Finance Department"
}

df_functions["department_name"] = (
    df_functions["department_name"].map(department_map)
)

print(df_functions)

# Filter columns containing "name"
name_columns = df.filter(like="name")
print(name_columns)
```

Phase 10: Pivot Tables
Task 24: Mean and Maximum Salary by Department

```python

pivot_result = pd.pivot_table(
    df,
    index="department_name",
    values="salary",
    aggfunc=["mean", "max"]
)

print(pivot_result)
```

Phase 11: Boolean Filtering
Task 25: Boolean Masking

```python

result = df[
    (
        (df["age"] < 35) &
        (df["salary"] > 55000)
    ) |
    (df["department_name"] == "IT")
]

print(result)
```

Phase 12: Date and Time

```python

dates = [
    "2020-01-15",
    "2021-03-20",
    "2019-06-10",
    "2022-08-05",
    "2020-11-12"
]

df_dates = df.copy()

# Add hire_date; remaining employees receive missing dates
df_dates["hire_date"] = pd.NaT
df_dates.loc[df_dates.index[:5], "hire_date"] = dates

# Convert to datetime
df_dates["hire_date"] = pd.to_datetime(df_dates["hire_date"])

# Extract year
df_dates["hire_year"] = df_dates["hire_date"].dt.year

# Extract month name
df_dates["hire_month"] = df_dates["hire_date"].dt.month_name()

# Calculate days since joining
today = pd.Timestamp.today().normalize()
df_dates["days_since_joining"] = (
    today - df_dates["hire_date"]
).dt.days

print(df_dates[
    ["name", "hire_date", "hire_year",
     "hire_month", "days_since_joining"]
])
```
